```python
# sistema de caixa eletronico
def depositar(saldo, valor_deposito):
  # fazer o algoritmo do deposito (saldo + deposito)
  novo_saldo = saldo + valor_deposito
  # informar ao usuario o saldo atual e o novo
  print(f"Saldo anterior: R$ {saldo:.2f}")
  print(f"Depósito de: R$ {valor_deposito:.2f}")
  print(f"Novo saldo: R$ {novo_saldo:.2f}")
  return novo_saldo

def sacar(saldo, valor_saque):
  # opcional - se o saldo for menor que o saque, nao permitir
  if valor_saque > saldo:
    print("Saldo insuficiente para realizar o saque!")
    return saldo

  # fazer o algoritmo do saque (saldo - saque)
  novo_saldo = saldo - valor_saque
  print(f"Saldo anterior: R$ {saldo:.2f}")
  print(f"Saque de: R$ {valor_saque:.2f}")
  print(f"Novo saldo: R$ {novo_saldo:.2f}")
  return novo_saldo
```

```python
# Definições das funções
def depositar(saldo, valor_deposito):
    novo_saldo = saldo + valor_deposito
    print(f"Depósito de R$ {valor_deposito:.2f} realizado com sucesso!")
    print(f"Seu novo saldo é: R$ {novo_saldo:.2f}")
    return novo_saldo

def sacar(saldo, valor_saque):
    if valor_saque > saldo:
        print("Saldo insuficiente para realizar este saque.")
        return saldo
    novo_saldo = saldo - valor_saque
    print(f"Saque de R$ {valor_saque:.2f} realizado com sucesso!")
    print(f"Seu novo saldo é: R$ {novo_saldo:.2f}")
    return novo_saldo

# Fluxo principal do programa
print("Bem vindo ao GEN-CAIXA")
saldo = float(input("Informe seu saldo: "))
print("Escolha uma das opções:")
print("1 - ver saldo")
print("2 - depositar")
print("3 - sacar")
print("qualquer outra opção - sair")
opcao = input("Digite a opção desejada: ")

if opcao == "1":
    print("Seu saldo atual é: R$", saldo)

elif opcao == "2":
    valor = float(input("Qual valor você quer depositar? "))
    saldo = depositar(saldo, valor)

elif opcao == "3":
    valor = float(input("Qual valor você quer sacar? "))
    saldo = sacar(saldo, valor)
else:
    print("Saindo do sistema. Até logo!")
```
